```python
import pandas as pd
```

```python
df = pd.read_csv("vgsales.csv")
```

**Create null value and clean dataframes**

Find all rows that have null values. Clean the base dataframe

```python
null_data = df[df.isnull().any(axis=1)]
```

```python
clean_data = df.dropna()
```


Explore the dataframe of rows with null values

```python
null_data.shape
```

```python
clean_data.shape
```

```python
null_data.head()
```

307 rows have missing values, 16291 do not

**Question: Do rows with missing values contribute to the statistics we're looking at?**

```python
null_data['Global_Sales'].describe()
```

```python
clean_data['Global_Sales'].describe()
```

The rows with missing values probably do not contribute much to overall statistics.

**Question: Are some genres overly impacted by missing values?**

```python
#unique(), describe() and info do not give counts of unique values
null_data['Genre'].value_counts()
```

```python
clean_data['Genre'].value_counts()
```

```python
null_data['Publisher'].value_counts().head(10)
```

```python
clean_data['Publisher'].value_counts().head(10)
```

No genre is specifically impacted by missing values

***Conclusion***

If we want to know if a specific game influenced the popularity of a genre, we should ignore missing values. They do not contribute significantly to sales, and are missing the Year value whi
